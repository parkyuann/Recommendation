```python
import pandas as pd

df = pd.read_csv('C:/Users/user/OneDrive/Desktop/새 폴더/data/processed/0507_userindex.csv')
mart = pd.read_csv('C:/Users/user/OneDrive/Desktop/새 폴더/data/processed/vod_mart_processed.csv')
```

```python
df.drop(columns = {'sha2_hash', 'user_label', 'asset_nm'}, inplace = True)
```

```python
# mart에서 필요한 컬럼만 선택
mart_subset = mart[['asset_id', 'super_asset_nm']]

# df와 mart_subset을 asset_id를 기준으로 병합
df = pd.merge(df, mart_subset, on='asset_id', how='left')

# 결과 확인
print(f"병합 후 데이터 크기: {df.shape}")
print(f"super_asset_nm 컬럼의 결측치 수: {df['super_asset_nm'].isnull().sum()}")
```

```python
# mart에서 필요한 컬럼만 선택
mart_subset = mart[['asset_id', 'category_l1', 'category_l2']]

# df와 mart_subset을 asset_id를 기준으로 병합
df = pd.merge(df, mart_subset, on='asset_id', how='left')

# 결과 확인
print(f"병합 후 데이터 크기: {df.shape}")
print(f"category_l1 컬럼의 결측치 수: {df['category_l1'].isnull().sum()}")
print(f"category_l2 컬럼의 결측치 수: {df['category_l2'].isnull().sum()}")
```

```python
# mart에서 필요한 컬럼만 선택
mart_subset = mart[['asset_id', 'genre']]

# df와 mart_subset을 asset_id를 기준으로 병합
df = pd.merge(df, mart_subset, on='asset_id', how='left')

# 결과 확인
print(f"병합 후 데이터 크기: {df.shape}")
print(f"genre 컬럼의 결측치 수: {df['genre'].isnull().sum()}")
```
